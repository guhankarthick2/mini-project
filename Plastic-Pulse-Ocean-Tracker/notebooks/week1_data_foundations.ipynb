# Week 1 — Data Foundations & Image Pre-processing
**Plastic-Pulse Ocean Tracker**

This notebook walks through the Week 1 deliverable:
1. Inspect the curated balanced dataset
2. Verify 224×224 resize + [0, 1] normalization
3. Review augmentation variants (flip / rotation)

```python
from pathlib import Path
import json
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

ROOT = Path('..').resolve()
RAW = ROOT / 'data' / 'raw'
PROCESSED = ROOT / 'data' / 'processed'
AUGMENTED = ROOT / 'data' / 'augmented'
OUTPUTS = ROOT / 'outputs'
CLASSES = ['marine_life', 'plastic_debris']

print('Project root:', ROOT)
```

## 1. Balanced dataset check

```python
def count_images(folder: Path) -> int:
    return len([p for p in folder.glob('*') if p.suffix.lower() in {'.jpg', '.jpeg', '.png'}])

for name in CLASSES:
    n = count_images(RAW / name)
    print(f'{name:16s}: {n} raw images')

assert count_images(RAW / 'marine_life') == count_images(RAW / 'plastic_debris'), 'Dataset is not balanced'
print('Dataset is balanced.')
```

## 2. Pre-processing verification (224×224, normalized)

```python
sample = next((PROCESSED / 'marine_life').glob('*.npy'))
arr = np.load(sample)
print('shape:', arr.shape)
print('dtype:', arr.dtype)
print('min/max:', float(arr.min()), float(arr.max()))
assert arr.shape == (224, 224, 3)
assert arr.dtype == np.float32
assert 0.0 <= arr.min() and arr.max() <= 1.0
print('Pre-processing checks passed.')
```

## 3. Augmentation gallery

```python
fig, axes = plt.subplots(2, 6, figsize=(14, 5))
for row, name in enumerate(CLASSES):
    samples = sorted((AUGMENTED / name).glob('*__*.jpg'))[:6]
    for col, path in enumerate(samples):
        axes[row, col].imshow(Image.open(path))
        axes[row, col].set_title(path.stem.split('__')[-1], fontsize=8)
        axes[row, col].axis('off')
plt.suptitle('Augmented samples ready for training')
plt.tight_layout()
plt.show()
```
