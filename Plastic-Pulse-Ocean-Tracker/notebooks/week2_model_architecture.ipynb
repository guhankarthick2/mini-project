# Week 2 — Pretrained Feature Extractor + Custom Head
**Plastic-Pulse Ocean Tracker**

This notebook builds a binary classifier (**Marine Life** vs **Plastic Debris**) using:
1. A pretrained CNN (`MobileNetV2` or `ResNet50`) as a **frozen feature extractor**
2. A **custom classification head**
3. Compile + architecture inspection

```python
import sys
from pathlib import Path

import tensorflow as tf

ROOT = Path('..').resolve()
sys.path.insert(0, str(ROOT))

from models.classifier import build_classifier, compile_model, inspect_model

print('TensorFlow', tf.__version__)
```

## 1. Build model (frozen MobileNetV2 + custom head)

```python
model = build_classifier(backbone='mobilenetv2', dense_units=128, dropout=0.3)
model = compile_model(model, learning_rate=1e-3)
model
```

## 2. Inspect architecture

```python
arch = inspect_model(model, ROOT / 'outputs' / 'week2')
arch['params']
```

## 3. Confirm backbone is frozen

```python
for layer in model.layers:
    print(f'{layer.name:30s} trainable={layer.trainable}')
```
